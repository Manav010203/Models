```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
df= pd.read_csv("Dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.head()
```

```python
def clean_text(text):
    return text.lower()
df['clean_text'] = df['selected_text'].apply(clean_text)

print(df[['selected_text', 'clean_text']].head())
```

```python
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df['clean_text'],df['sentiment'],test_size=0.2, random_state=42)
X_test.head()
```

```python
vectorizer = TfidfVectorizer(max_features=5000,ngram_range=(1,2))
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("TF-IDF shape (train):", X_train_tfidf.shape)
print("TF-IDF shape (test):", X_test_tfidf.shape)
```

```python
bnb = BernoulliNB()
bnb.fit(X_train_tfidf,y_train)
pred = bnb.predict(X_test_tfidf)

print(f"accuracy of bnb:{accuracy_score(y_test,pred)}")
print(f"classification report:{classification_report(y_test,pred)}")
```

```python
svm = LinearSVC()
svm.fit(X_train_tfidf,y_train)
pred_svm = svm.predict(X_test_tfidf)


print(f"accuracy of bnb:{accuracy_score(y_test,pred_svm)}")
print(f"classification report:{classification_report(y_test,pred_svm)}")
```

```python
LR = LogisticRegression()
LR.fit(X_train_tfidf,y_train)
pred_lr = LR.predict(X_test_tfidf)


print(f"accuracy of bnb:{accuracy_score(y_test,pred_lr)}")
print(f"classification report:{classification_report(y_test,pred_lr)}")
```

```python
# making perdictions
sample_tweets = ["I love this!", "I hate that!", "It was okay, not great.","neutral"]
vectorize_sample = vectorizer.transform(sample_tweets)

print("prediction on sample tweet\n")
print(f"bnb :{bnb.predict(vectorize_sample)}")
print(f"svm : {svm.predict(vectorize_sample)}")
print(f"lr : {LR.predict(vectorize_sample)}")
```
